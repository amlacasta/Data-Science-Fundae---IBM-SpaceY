## 1. Objetivo del caso

El caso SpaceY trabaja con datos de lanzamientos de SpaceX Falcon 9 y plantea una tarea de clasificación:

```text
Class = 1 → primera etapa aterriza correctamente
Class = 0 → primera etapa no aterriza correctamente
```


## 2. Flujo trabajado

El curso permitió practicar un flujo bastante completo de ciencia de datos:

1. Obtención de datos desde API.
2. Extracción de datos con web scraping.
3. Limpieza y transformación de variables.
4. Análisis exploratorio.
5. Consultas SQL.
6. Entrenamiento de modelos supervisados.
7. Comparación de resultados.


````python
import pandas as pd

models = pd.DataFrame({
    'model': ['Logistic Regression', 'SVM', 'KNN', 'Decision Tree'],
    'test_accuracy': [0.8333, 0.8333, 0.8333, 0.7222]
})

models.sort_values('test_accuracy', ascending=False)
````

## 3. Resultado principal

Los modelos Logistic Regression, SVM y KNN empatan con una accuracy de test de `0.8333`. Decision Tree obtiene `0.7222`.

La interpretación debe hacerse con prudencia porque el conjunto de test es pequeño. El objetivo principal del ejercicio es aprender el flujo de trabajo, no obtener un modelo productivo.


````python
import matplotlib.pyplot as plt

ax = models.sort_values('test_accuracy').plot(
    kind='barh',
    x='model',
    y='test_accuracy',
    legend=False,
    figsize=(8, 4)
)
ax.set_title('Comparativa de modelos — Accuracy en test')
ax.set_xlabel('Accuracy')
ax.set_ylabel('Modelo')
plt.xlim(0, 1)
plt.show()
````

## 4. Aprendizajes clave

- Una API permite construir datasets desde fuentes vivas.
- El web scraping ayuda a complementar información cuando los datos están en tablas HTML.
- El data wrangling es decisivo para convertir datos brutos en datos utilizables.
- SQL sigue siendo esencial para responder preguntas analíticas.
- En datasets pequeños, modelos sencillos pueden rendir tan bien como modelos más complejos.
- La evaluación debe contextualizarse siempre: tamaño del test, calidad de datos y objetivo del análisis.
